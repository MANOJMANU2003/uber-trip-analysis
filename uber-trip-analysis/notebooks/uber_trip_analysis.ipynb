```python
median_distance = df["Trip_Distance_km"].median()
df["Trip_Distance_km"] = df["Trip_Distance_km"].fillna(median_distance)

print("Median distance used:", median_distance)
print("Missing values remaining:", df["Trip_Distance_km"].isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print("Duplicate Trip IDs:", df["Trip_ID"].duplicated().sum())
```

```python
print("Distance <= 0:", (df["Trip_Distance_km"] <= 0).sum())
print("Fare <= 0:", (df["Fare_Amount"] <= 0).sum())
print("Surge <= 0:", (df["Surge_Multiplier"] <= 0).sum())
print("Total Fare <= 0:", (df["Total_Fare"] <= 0).sum())
```

```python
df["Pickup_Date"] = pd.to_datetime(df["Pickup_Date"],format="mixed",dayfirst=True)
print(df["Pickup_Date"].dtype)
```

```python
print(df["Pickup_Date"].head())
print(df["Pickup_Date"].tail())
```

```python
print("Data type:", df["Pickup_Date"].dtype)

print("\n First 5 Dates:")
print(df["Pickup_Date"].head())

print("\n Date range:")
print(df["Pickup_Date"].min())
print(df["Pickup_Date"].max())
```

```python
df["Pickup_Date"].isnull().sum()
```

```python
df["Hour"] = df["Pickup_Date"].dt.hour
df["Day"] = df["Pickup_Date"].dt.day
df["Day_Name"] = df["Pickup_Date"].dt.day_name()
df["Month"] = df["Pickup_Date"].dt.month
df["Month_Name"] = df["Pickup_Date"].dt.month_name()
df["Year"] = df["Pickup_Date"].dt.year
df["Is_Weekend"] = df["Pickup_Date"].dt.dayofweek>=5
```

```python
df.head()
```

```python
df[["Pickup_Date","Hour","Day_Name","Is_Weekend"]].head(10)
```

```python
hourly_trips = df["Hour"].value_counts().sort_index()
print(hourly_trips)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

sns.barplot(x = hourly_trips.index, y = hourly_trips.values)

plt.title("Uber Trip by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Trips")
plt.xticks(range(24))

plt.show()
```

```python
busiest_hour = hourly_trips.idxmax()
max_trips = hourly_trips.max()

print("Busiest Hour:", busiest_hour)
print("Number of Trips:", max_trips)
```

```python
quietest_hour = hourly_trips.idxmin()
min_trips = hourly_trips.min()

print("Quietest Hour:", quietest_hour)
print("Number of Trips:", min_trips)
```

```python
weekend_trips = df["Is_Weekend"].value_counts()
print(weekend_trips)
```

```python
weekend_summary =df.groupby("Is_Weekend").size()
print(weekend_summary)
```

```python
weekend_summary.index = ["Weekday"if x == False else "Weekend" for x in weekend_summary.index]
print(weekend_summary)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x = weekend_summary.index,
    y = weekend_summary.values)

plt.title("Uber Trips: Weekday vs Weekend")
plt.xlabel("Day Type")
plt.ylabel("Number of Trips")

plt.show()
```

```python
day_trips = df["Day_Name"].value_counts()
day_order = [
    "Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"
]
day_trips = day_trips.reindex(day_order)
print(day_trips)
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x = day_trips.index,
    y = day_trips.values)

plt.title("Uber Trips  by Day of Week")
plt.xlabel("Day")
plt.ylabel("Number of Trips")
plt.xticks(rotation=30)

plt.show()
```

```python
total_revenue = df["Total_Fare"].sum()
print("Total Revenue:",total_revenue)
```

```python
average_fare = df["Total_Fare"].mean()
print("Average Fare:",round(average_fare, 2))
```

```python
revenue_by_ride = (
    df.groupby("Ride_Type")["Total_Fare"]
    .sum()
    .sort_values(ascending=False)
)
print(revenue_by_ride)
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x = revenue_by_ride.index,
    y = revenue_by_ride.values)

plt.title("Revenue by Ride Type")
plt.xlabel("Ride Type")
plt.ylabel("Total Revenue")
plt.xticks(rotation=30)

plt.show()
```

```python
trips_by_ride = df["Ride_Type"].value_counts()
print(trips_by_ride)
```

```python
avg_revenue_by_ride = (
    df.groupby("Ride_Type")["Total_Fare"]
    .mean()
    .sort_values(ascending=False)
)
print(avg_revenue_by_ride.round(2))
```

```python
print("Average Fare Amount:",
     round(df["Fare_Amount"].mean(), 2))

print("Average Total Amount:",
     round(df["Total_Fare"].mean(), 2))
```

```python
payment_revenue = (
    df.groupby("Payment_Type")["Total_Fare"]
    .agg(["sum","count","mean"])
    .sort_values("sum",ascending=False)
)
payment_revenue
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x=payment_revenue.index,
    y=payment_revenue["sum"]
)
plt.title("Revenue by Payment Type")
plt.xlabel("Payment Type")
plt.ylabel("Total Revenue")

plt.show()
```

```python
plt.figure(figsize=(8, 8))

plt.pie(
    payment_revenue["count"],
    labels=payment_revenue.index,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops=dict(width=0.4)
)

plt.title("Trip Distribution by Payment Type")

plt.show()
```

```python
pickup_trips = df["Pickup_Location"].value_counts()
print(pickup_trips)
```

```python
pickup_revenue = (
    df.groupby("Pickup_Location")["Total_Fare"]
    .agg(["sum","count","mean"])
    .sort_values("sum", ascending=False)
)
pickup_revenue
```

```python
top_pickup = pickup_revenue.head(10)
print(top_pickup)
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    top_pickup.index,
    top_pickup["sum"]
)

plt.title("Revenue by Pickup Location")
plt.xlabel("Total Revenue")
plt.ylabel("Pickup Location")

plt.gca().invert_yaxis()

plt.show()
```

```python
dropoff_revenue = (
    df.groupby("Dropoff_Location")["Total_Fare"]
      .agg(["sum","count","mean"])
      .sort_values("sum",ascending=False)
)
dropoff_revenue
```

```python
top_dropoff = dropoff_revenue.head(10)
print(top_dropoff)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_dropoff.index,
    y=top_dropoff["sum"]
)

plt.title("Top Dropoff Locations by Revenue")
plt.xlabel("Dropoff Location")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)

plt.show()
```

```python
route_analysis = (
    df.groupby(["Pickup_Location", "Dropoff_Location"])["Total_Fare"]
      .agg(["sum", "count", "mean"])
      .sort_values("sum", ascending=False)
)

route_analysis.head(10)
```

```python
df["Surge_Multiplier"].value_counts().sort_index()
```

```python
surge_analysis = (
    df.groupby("Surge_Multiplier")
      .agg(
          Trips=("Trip_ID", "count"),
          Average_Fare=("Total_Fare", "mean"),
          Total_Revenue=("Total_Fare", "sum"),
          Average_Distance=("Trip_Distance_km", "mean")
    )
    .sort_index()
)

surge_analysis.round(2)
```

```python
surge_correlation = df["Surge_Multiplier"].corr(df["Total_Fare"])

print("Surge vs Total Fare correlation:",
      round(surge_correlation, 3))
```

```python
daily_revenue = (
    df.groupby(df["Pickup_Date"].dt.date)["Total_Fare"]
      .agg(["sum", "count", "mean"])
)

daily_revenue.round(2)
```

```python
highest_revenue_day = daily_revenue["sum"].idxmax()
highest_revenue = daily_revenue["sum"].max()

print("Highest Revenue Day:", highest_revenue_day)
print("Revenue:", round(highest_revenue, 2))
```

```python
lowest_revenue_day = daily_revenue["sum"].idxmin()
lowest_revenue = daily_revenue["sum"].min()

print("Lowest Revenue Day:", lowest_revenue_day)
print("Revenue:", round(lowest_revenue, 2))
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    daily_revenue.index,
    daily_revenue["sum"]
)

plt.title("Daily Uber Revenue")
plt.xlabel("Date")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
